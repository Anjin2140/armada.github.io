````python
# MAI C# Local LLM Engine

A modular C# framework for routing prompts through local LLM backends, deterministic skills, session memory, evidence rules, and failure handling.

This engine treats AI behaviors as isolated, composable skills with strict contracts, boundaries, inputs, outputs, dependencies, and execution rules.

---

## 1. Architecture

```text
MAI Runtime
  ├─ App Host
  ├─ LLM Backend
  │    ├─ Ollama Backend
  │    ├─ LLamaSharp Backend
  │    └─ Future Backend Adapters
  │
  ├─ Session Manager
  ├─ Skill Registry
  ├─ Skill Router
  ├─ Skill Executor
  ├─ Evidence Manager
  ├─ Failure Handler
  └─ Console / UI / API Layer
```

---

## 2. Project Structure

```text
MaiEngine/
  Program.cs
  Core/
    ChatMessage.cs
    ChatSession.cs
    GenerationOptions.cs
    LlmResult.cs
  Backends/
    ILlmBackend.cs
    OllamaBackend.cs
  Skills/
    SkillDefinition.cs
    SkillRegistry.cs
    SkillRouter.cs
    SkillExecutor.cs
    SkillExecutionResult.cs
  Evidence/
    EvidenceManager.cs
  Failure/
    FailureHandler.cs
```

---

## 3. Core Models

### `Core/ChatMessage.cs`

```csharp
namespace MaiEngine.Core;

public sealed record ChatMessage(
    string Role,
    string Content
);
```

### `Core/ChatSession.cs`

```csharp
namespace MaiEngine.Core;

public sealed class ChatSession
{
    public Guid SessionId { get; } = Guid.NewGuid();

    public List<ChatMessage> Messages { get; } = new();

    public Dictionary<string, object> State { get; } = new();

    public void AddSystem(string content)
        => Messages.Add(new ChatMessage("system", content));

    public void AddUser(string content)
        => Messages.Add(new ChatMessage("user", content));

    public void AddAssistant(string content)
        => Messages.Add(new ChatMessage("assistant", content));
}
```

### `Core/GenerationOptions.cs`

```csharp
namespace MaiEngine.Core;

public sealed record GenerationOptions(
    double Temperature = 0.7,
    double TopP = 0.9,
    int MaxTokens = 1024,
    bool Stream = true
);
```

### `Core/LlmResult.cs`

```csharp
namespace MaiEngine.Core;

public sealed record LlmResult(
    bool Success,
    string Output,
    IReadOnlyList<string> Warnings
);
```

---

## 4. LLM Backend Interface

### `Backends/ILlmBackend.cs`

```csharp
using MaiEngine.Core;

namespace MaiEngine.Backends;

public interface ILlmBackend
{
    IAsyncEnumerable<string> GenerateAsync(
        ChatSession session,
        GenerationOptions options,
        CancellationToken ct = default);
}
```

---

## 5. Ollama Backend

### `Backends/OllamaBackend.cs`

```csharp
using System.Net.Http.Json;
using System.Runtime.CompilerServices;
using System.Text.Json;
using System.Text.Json.Serialization;
using MaiEngine.Core;

namespace MaiEngine.Backends;

public sealed class OllamaBackend : ILlmBackend
{
    private readonly HttpClient _http;
    private readonly string _model;

    public OllamaBackend(
        string model,
        string baseUrl = "http://localhost:11434")
    {
        _model = model;
        _http = new HttpClient
        {
            BaseAddress = new Uri(baseUrl)
        };
    }

    public async IAsyncEnumerable<string> GenerateAsync(
        ChatSession session,
        GenerationOptions options,
        [EnumeratorCancellation] CancellationToken ct = default)
    {
        var request = new
        {
            model = _model,
            stream = options.Stream,
            messages = session.Messages.Select(m => new
            {
                role = m.Role,
                content = m.Content
            }),
            options = new
            {
                temperature = options.Temperature,
                top_p = options.TopP,
                num_predict = options.MaxTokens
            }
        };

        using var response = await _http.PostAsJsonAsync("/api/chat", request, ct);
        response.EnsureSuccessStatusCode();

        await using var stream = await response.Content.ReadAsStreamAsync(ct);
        using var reader = new StreamReader(stream);

        while (!reader.EndOfStream && !ct.IsCancellationRequested)
        {
            var line = await reader.ReadLineAsync(ct);

            if (string.IsNullOrWhiteSpace(line))
                continue;

            var chunk = JsonSerializer.Deserialize<OllamaChunk>(line);

            if (!string.IsNullOrEmpty(chunk?.Message?.Content))
                yield return chunk.Message.Content;

            if (chunk?.Done == true)
                yield break;
        }
    }

    private sealed class OllamaChunk
    {
        [JsonPropertyName("message")]
        public OllamaMessage? Message { get; set; }

        [JsonPropertyName("done")]
        public bool Done { get; set; }
    }

    private sealed class OllamaMessage
    {
        [JsonPropertyName("content")]
        public string? Content { get; set; }
    }
}
```

---

## 6. Skill Definition Layer

### `Skills/SkillDefinition.cs`

```csharp
namespace MaiEngine.Skills;

public sealed record SkillDefinition(
    string SkillId,
    string Trigger,
    string Purpose,
    SkillState State,
    SkillBoundaries Boundaries,
    SkillDataContracts DataContracts,
    SkillLogicGates LogicGates,
    IReadOnlyList<string> ExecutionSequence,
    IReadOnlyList<SkillDependency> Dependencies,
    SkillFailureHandling FailureHandling
);

public enum SkillState
{
    ConfirmedBehavior,
    DesiredFutureBehavior
}

public sealed record SkillBoundaries(
    string WhenToUse,
    string WhenNotToUse,
    string SafetyLimits
);

public sealed record SkillDataContracts(
    IReadOnlyList<SkillInputField> Inputs,
    string ExpectedOutput,
    IReadOnlyList<string> EvidenceHierarchy
);

public sealed record SkillInputField(
    string Field,
    string Type,
    bool Required,
    string? Fallback
);

public sealed record SkillLogicGates(
    string AllowedAssumptions,
    string ProhibitedAssumptions,
    string DecisionRules,
    string TerminationRule
);

public sealed record SkillDependency(
    string SystemOrTool,
    bool Mandatory,
    string Fallback
);

public sealed record SkillFailureHandling(
    string MissingFields,
    string ContradictorySources,
    string ToolFailure
);
```

---

## 7. Skill Registry

### `Skills/SkillRegistry.cs`

```csharp
namespace MaiEngine.Skills;

public sealed class SkillRegistry
{
    private readonly List<SkillDefinition> _skills = new();

    public IReadOnlyList<SkillDefinition> Skills => _skills;

    public void Register(SkillDefinition skill)
    {
        if (_skills.Any(s => s.SkillId == skill.SkillId))
            throw new InvalidOperationException($"Skill already registered: {skill.SkillId}");

        _skills.Add(skill);
    }

    public SkillDefinition? GetById(string skillId)
    {
        return _skills.FirstOrDefault(s => s.SkillId == skillId);
    }
}
```

---

## 8. Skill Router

### `Skills/SkillRouter.cs`

```csharp
namespace MaiEngine.Skills;

public interface ISkillRouter
{
    SkillDefinition? SelectSkill(
        string userInput,
        IEnumerable<SkillDefinition> skills);
}

public sealed class SkillRouter : ISkillRouter
{
    public SkillDefinition? SelectSkill(
        string userInput,
        IEnumerable<SkillDefinition> skills)
    {
        var normalized = userInput.Trim();

        return skills
            .Where(skill =>
                normalized.Contains(skill.Trigger, StringComparison.OrdinalIgnoreCase) ||
                normalized.Contains(skill.Purpose, StringComparison.OrdinalIgnoreCase) ||
                normalized.Contains(skill.SkillId, StringComparison.OrdinalIgnoreCase))
            .OrderByDescending(skill => skill.State == SkillState.ConfirmedBehavior)
            .FirstOrDefault();
    }
}
```

---

## 9. Skill Execution Result

### `Skills/SkillExecutionResult.cs`

```csharp
namespace MaiEngine.Skills;

public sealed record SkillExecutionResult(
    bool Success,
    string Output,
    IReadOnlyList<string> Warnings
);
```

---

## 10. Evidence Manager

### `Evidence/EvidenceManager.cs`

```csharp
using MaiEngine.Skills;

namespace MaiEngine.Evidence;

public sealed class EvidenceManager
{
    public IReadOnlyList<string> ResolveEvidenceHierarchy(SkillDefinition skill)
    {
        return skill.DataContracts.EvidenceHierarchy;
    }

    public string BuildEvidenceInstruction(SkillDefinition skill)
    {
        var hierarchy = ResolveEvidenceHierarchy(skill);

        return string.Join(
            Environment.NewLine,
            hierarchy.Select((item, index) => $"{index + 1}. {item}")
        );
    }
}
```

---

## 11. Failure Handler

### `Failure/FailureHandler.cs`

```csharp
using MaiEngine.Skills;

namespace MaiEngine.Failure;

public sealed class FailureHandler
{
    public string HandleMissingFields(SkillDefinition skill)
    {
        return skill.FailureHandling.MissingFields;
    }

    public string HandleContradictorySources(SkillDefinition skill)
    {
        return skill.FailureHandling.ContradictorySources;
    }

    public string HandleToolFailure(SkillDefinition skill, Exception ex)
    {
        return $"{skill.FailureHandling.ToolFailure}: {ex.Message}";
    }
}
```

---

## 12. Skill Executor

### `Skills/SkillExecutor.cs`

```csharp
using MaiEngine.Backends;
using MaiEngine.Core;
using MaiEngine.Evidence;
using MaiEngine.Failure;

namespace MaiEngine.Skills;

public interface ISkillExecutor
{
    Task<SkillExecutionResult> ExecuteAsync(
        SkillDefinition skill,
        string userInput,
        CancellationToken ct = default);
}

public sealed class SkillExecutor : ISkillExecutor
{
    private readonly ILlmBackend _backend;
    private readonly EvidenceManager _evidenceManager;
    private readonly FailureHandler _failureHandler;

    public SkillExecutor(
        ILlmBackend backend,
        EvidenceManager evidenceManager,
        FailureHandler failureHandler)
    {
        _backend = backend;
        _evidenceManager = evidenceManager;
        _failureHandler = failureHandler;
    }

    public async Task<SkillExecutionResult> ExecuteAsync(
        SkillDefinition skill,
        string userInput,
        CancellationToken ct = default)
    {
        var warnings = new List<string>();

        try
        {
            var session = new ChatSession();

            session.AddSystem(BuildSystemPrompt(skill));
            session.AddUser(userInput);

            var output = "";

            await foreach (var token in _backend.GenerateAsync(
                session,
                new GenerationOptions(),
                ct))
            {
                output += token;
            }

            return new SkillExecutionResult(
                Success: true,
                Output: output,
                Warnings: warnings
            );
        }
        catch (Exception ex)
        {
            warnings.Add(_failureHandler.HandleToolFailure(skill, ex));

            return new SkillExecutionResult(
                Success: false,
                Output: "",
                Warnings: warnings
            );
        }
    }

    private string BuildSystemPrompt(SkillDefinition skill)
    {
        var evidence = _evidenceManager.BuildEvidenceInstruction(skill);

        return $"""
        You are executing a deterministic MAI skill.

        Skill ID:
        {skill.SkillId}

        Purpose:
        {skill.Purpose}

        Trigger:
        {skill.Trigger}

        State:
        {skill.State}

        Boundaries:
        When to use: {skill.Boundaries.WhenToUse}
        When not to use: {skill.Boundaries.WhenNotToUse}
        Safety limits: {skill.Boundaries.SafetyLimits}

        Expected output:
        {skill.DataContracts.ExpectedOutput}

        Evidence hierarchy:
        {evidence}

        Logic gates:
        Allowed assumptions: {skill.LogicGates.AllowedAssumptions}
        Prohibited assumptions: {skill.LogicGates.ProhibitedAssumptions}
        Decision rules: {skill.LogicGates.DecisionRules}
        Termination rule: {skill.LogicGates.TerminationRule}

        Execution sequence:
        {string.Join(Environment.NewLine, skill.ExecutionSequence.Select((x, i) => $"{i + 1}. {x}"))}

        Follow the skill contract exactly.
        Do not merge unrelated capabilities.
        Do not invent missing required fields.
        If required information is missing, follow the failure handling rule.
        """;
    }
}
```

---

## 13. Example Skill: Operational Skill Capture

### `Program.cs` registers this skill at startup.

```csharp
using MaiEngine.Skills;

public static class BuiltInSkills
{
    public static SkillDefinition OperationalSkillCapture()
    {
        return new SkillDefinition(
            SkillId: "System.Capture_Operational_Skill",
            Trigger: "Formalize the preceding workflow into a standard operational skill template.",
            Purpose: "Ingest user directives and formalize them into a strict, reproducible, deterministic AI operational skill template.",
            State: SkillState.ConfirmedBehavior,

            Boundaries: new SkillBoundaries(
                WhenToUse: "When a new repeatable workflow, analytical method, or system interaction is established and validated.",
                WhenNotToUse: "For one-off conversational queries, creative writing, or broad ambiguous brainstorming sessions.",
                SafetyLimits: "Cannot overwrite existing skills without explicit authorization; cannot combine multiple distinct capabilities into one capture."
            ),

            DataContracts: new SkillDataContracts(
                Inputs: new List<SkillInputField>
                {
                    new("RawWorkflowDescription", "String", true, null),
                    new("ExecutionParameters", "String", false, null)
                },
                ExpectedOutput: "Three artifacts: YAML Skill Draft, one-line Trigger Summary, Next-Session Gap List.",
                EvidenceHierarchy: new List<string>
                {
                    "Explicit user instructions in the current session.",
                    "Confirmed outputs from the immediately preceding interaction."
                }
            ),

            LogicGates: new SkillLogicGates(
                AllowedAssumptions: "Assume the user wants the YAML schema format unless another format is explicitly requested.",
                ProhibitedAssumptions: "Do not assume the success condition of the skill if the user has not defined it.",
                DecisionRules: "If the user request contains multiple distinct actions, isolate the primary action and push secondary actions to the Gap List.",
                TerminationRule: "Successful generation of the three required artifacts."
            ),

            ExecutionSequence: new List<string>
            {
                "Extract the core capability, trigger, inputs, and failure conditions from the user prompt.",
                "Map the extracted data into the standardized YAML Skill_ID schema.",
                "Print the completed schema, formulate the one-line trigger, and build the gap list."
            },

            Dependencies: new List<SkillDependency>
            {
                new("Standard Text Generation", false, "Return structured text only.")
            },

            FailureHandling: new SkillFailureHandling(
                MissingFields: "Populate missing schema fields with [USER DEFINITION REQUIRED].",
                ContradictorySources: "Prioritize the most recent instruction in the current prompt.",
                ToolFailure: "Report exact tool error and return partial structured output if possible."
            )
        );
    }
}
```

---

## 14. Main Runtime

### `Program.cs`

```csharp
using MaiEngine.Backends;
using MaiEngine.Core;
using MaiEngine.Evidence;
using MaiEngine.Failure;
using MaiEngine.Skills;

var modelName = "llama3.1:8b";

var backend = new OllamaBackend(modelName);
var evidenceManager = new EvidenceManager();
var failureHandler = new FailureHandler();

var registry = new SkillRegistry();
registry.Register(BuiltInSkills.OperationalSkillCapture());

var router = new SkillRouter();
var executor = new SkillExecutor(
    backend,
    evidenceManager,
    failureHandler
);

Console.WriteLine("MAI Runtime initialized.");
Console.WriteLine($"Backend model: {modelName}");
Console.WriteLine("Type 'exit' to quit.");
Console.WriteLine();

while (true)
{
    Console.Write(">> ");
    var input = Console.ReadLine();

    if (string.IsNullOrWhiteSpace(input))
        continue;

    if (input.Equals("exit", StringComparison.OrdinalIgnoreCase))
        break;

    var skill = router.SelectSkill(input, registry.Skills);

    if (skill is not null)
    {
        Console.WriteLine();
        Console.WriteLine($"[Skill selected: {skill.SkillId}]");
        Console.WriteLine();

        var result = await executor.ExecuteAsync(skill, input);

        if (!result.Success)
        {
            Console.WriteLine("[Execution failed]");
            foreach (var warning in result.Warnings)
                Console.WriteLine($"Warning: {warning}");

            continue;
        }

        Console.WriteLine(result.Output);
        Console.WriteLine();
        continue;
    }

    var session = new ChatSession();

    session.AddSystem("""
    You are the default MAI assistant.
    Be direct, structured, and useful.
    Do not invent facts.
    Ask for missing required inputs when needed.
    """);

    session.AddUser(input);

    await foreach (var token in backend.GenerateAsync(
        session,
        new GenerationOptions()))
    {
        Console.Write(token);
    }

    Console.WriteLine();
    Console.WriteLine();
}
```

---

## 15. `.csproj`

```xml
<Project Sdk="Microsoft.NET.Sdk">

  <PropertyGroup>
    <OutputType>Exe</OutputType>
    <TargetFramework>net8.0</TargetFramework>
    <Nullable>enable</Nullable>
    <ImplicitUsings>enable</ImplicitUsings>
  </PropertyGroup>

</Project>
```

---

## 16. Running Locally

Install Ollama and pull a model:

```bash
ollama pull llama3.1:8b
```

Run the Ollama service:

```bash
ollama serve
```

Run the app:

```bash
dotnet run
```

---

## 17. Next Extensions

```text
1. Add persistent skill storage as YAML or JSON.
2. Add embedding-based skill routing instead of string matching.
3. Add session memory with SQLite.
4. Add tool adapters for filesystem, web, local documents, and vector search.
5. Add role-based permission boundaries per skill.
6. Add telemetry for skill selection, execution time, and failures.
7. Add a REST API or Avalonia/WPF front end.
```

---

## 18. Safety and Boundary Layer

The engine should preserve user control and privacy while still enforcing operational boundaries.

Recommended runtime policies:

```text
- Never execute destructive filesystem actions without explicit confirmation.
- Never overwrite skills unless the user explicitly authorizes replacement.
- Never invent missing required fields.
- Separate facts, inferences, and unknowns.
- Prefer local execution for private data.
- Log tool failures exactly.
- Keep each skill narrow and deterministic.
```

---

## 19. Bottom Line

C# should act as the deterministic control plane.

The LLM generates language.

The C# engine governs:

```text
- routing
- skill contracts
- state
- memory
- dependencies
- evidence
- failure handling
- permissions
- execution boundaries
```

This keeps the MAI ecosystem modular, inspectable, and scalable.
````

# 8. Skill Blueprint Schema

Use this YAML shape for ingesting new skills.

```yaml
Skill_ID: "[Namespace.Narrow_Capability_Name]"
```

```csharp
                normalized.Contains(skill.Purpose, StringComparison.OrdinalIgnoreCase) ||

    }
```

# 14. Skill Executor

```csharp
using MaiEngine.Backends;
using MaiEngine.Core;
using MaiEngine.Evidence;
using MaiEngine.Failure;

namespace MaiEngine.Skills;

public interface ISkillExecutor
{
    Task<SkillExecutionResult> ExecuteAsync(
        SkillDefinition skill,
        string userInput,
        CancellationToken ct = default);
}

public sealed class SkillExecutor : ISkillExecutor
{
    private readonly ILlmBackend _backend;
    private readonly EvidenceManager _evidenceManager;
    private readonly FailureHandler _failureHandler;

    public SkillExecutor(
        ILlmBackend backend,
        EvidenceManager evidenceManager,
        FailureHandler failureHandler)
    {
        _backend = backend;
        _evidenceManager = evidenceManager;
        _failureHandler = failureHandler;
    }

    public async Task<SkillExecutionResult> ExecuteAsync(
        SkillDefinition skill,
        string userInput,
        CancellationToken ct = default)
    {
        var warnings = new List<string>();

        try
        {
            var session = new ChatSession();

            session.AddSystem(BuildSystemPrompt(skill));
            session.AddUser(userInput);

            var output = "";

            await foreach (var token in _backend.GenerateAsync(
                session,
                new GenerationOptions(),
                ct))
            {
                output += token;
            }

            return new SkillExecutionResult(
                Success: true,
                Output: output,
                Warnings: warnings
            );
        }
        catch (Exception ex)
        {
            warnings.Add(_failureHandler.HandleToolFailure(skill, ex));

            return new SkillExecutionResult(
                Success: false,
                Output: "",
                Warnings: warnings
            );
        }
    }

    private string BuildSystemPrompt(SkillDefinition skill)
    {
        var evidence = _evidenceManager.BuildEvidenceInstruction(skill);

        return $"""
        You are executing a deterministic MAI skill.

        Skill ID:
        {skill.SkillId}

        Purpose:
        {skill.Purpose}

        Trigger:
        {skill.Trigger}

        State:
        {skill.State}

        Boundaries:
        When to use: {skill.Boundaries.WhenToUse}
        When not to use: {skill.Boundaries.WhenNotToUse}
        Safety limits: {skill.Boundaries.SafetyLimits}

        Expected output:
        {skill.DataContracts.ExpectedOutput}

        Evidence hierarchy:
        {evidence}

        Prohibited assumptions: {skill.LogicGates.ProhibitedAssumptions}
        Decision rules: {skill.LogicGates.DecisionRules}
        Termination rule: {skill.LogicGates.TerminationRule}

        Execution sequence:
        {string.Join(Environment.NewLine, skill.ExecutionSequence.Select((x, i) => $"{i + 1}. {x}"))}

        Follow the skill contract exactly.
        Do not merge unrelated capabilities.
        Do not invent missing required fields.
        If required information is missing, follow the failure handling rule.
        """;
    }
}
```

## 16.2 In-Memory Store

```csharp
namespace MaiEngine.Memory;

public sealed class InMemoryStore : IMemoryStore
{
    private readonly Dictionary<string, string> _store = new();

    public Task SaveAsync(string key, string value, CancellationToken ct = default)
    {
        _store[key] = value;
        return Task.CompletedTask;
    }

    public Task<string?> GetAsync(string key, CancellationToken ct = default)
    {
        _store.TryGetValue(key, out var value);
        return Task.FromResult(value);
    }

    public Task<IReadOnlyList<string>> SearchAsync(string query, CancellationToken ct = default)
    {
        var results = _store
            .Where(x => x.Key.Contains(query, StringComparison.OrdinalIgnoreCase) ||
                        x.Value.Contains(query, StringComparison.OrdinalIgnoreCase))
            .Select(x => x.Value)
            .ToList();

        return Task.FromResult<IReadOnlyList<string>>(results);
    }
}
```

```text
2. Do not retrieve memory unless routed.
3. Do not treat memory as absolute truth.
4. Use explicit current-session instructions over stored memory.
5. If memory conflicts with current user input, use current user input.
```

```csharp
    You are the default MAI assistant.
    Be direct, structured, and useful.
    Do not invent facts.
    Ask for missing required inputs when needed.
    """);

    session.AddUser(input);

    {
        Console.Write(token);
```

```python
import pandas as pd
from pathlib import Path

def process_csv_chunked(
    path: str,
    chunk_size: int = 50_000,
    required_columns=None
):
    summaries = []

    for index, chunk in enumerate(pd.read_csv(path, chunksize=chunk_size)):
        if required_columns:
            chunk = chunk[required_columns]

        summary = {
            "chunk_id": index,
            "rows": len(chunk),
            "columns": list(chunk.columns),
            "null_counts": chunk.isna().sum().to_dict()
        }

        summaries.append(summary)

    return {
        "total_chunks": len(summaries),
        "summaries": summaries
    }
```

```csharp
    public async Task<CoderResult> ExecuteAsync(
        CoderTask task,
        CancellationToken ct = default)
    {

                false,

            );
        }

        return await module.ExecuteAsync(task, ct);
    }

    private ICoderModule? SelectModule(string intent)
    {
        return _modules.FirstOrDefault(m =>
            intent.Contains(m.Name, StringComparison.OrdinalIgnoreCase));
    }
}
```

```csharp
public sealed class SemanticSkillRouter : ISemanticSkillRouter
{
    private readonly IEmbeddingProvider _embeddingProvider;
    private readonly ISkillEmbeddingStore _embeddingStore;
    private readonly double _threshold;

    public SemanticSkillRouter(
        IEmbeddingProvider embeddingProvider,
        ISkillEmbeddingStore embeddingStore,
        double threshold = 0.78)
    {
        _embeddingProvider = embeddingProvider;
        _embeddingStore = embeddingStore;
        _threshold = threshold;
    }

    public async Task<SkillDefinition?> SelectSkillAsync(
        string userInput,
        IEnumerable<SkillDefinition> skills,
        CancellationToken ct = default)
    {
        var inputEmbedding = await _embeddingProvider.EmbedAsync(userInput, ct);

        SkillDefinition? bestSkill = null;
        var bestScore = 0.0;

        foreach (var skill in skills)
        {
            var triggerEmbedding = await _embeddingStore.GetOrCreateAsync(
                skill.SkillId,
                skill.Trigger + "\n" + skill.Purpose,
                ct);

            var score = CosineSimilarity(inputEmbedding, triggerEmbedding);

            if (score > bestScore)
            {
                bestScore = score;
                bestSkill = skill;
            }
        }

        return bestScore >= _threshold ? bestSkill : null;
    }

    private static double CosineSimilarity(float[] a, float[] b)
    {
        if (a.Length != b.Length)
            throw new InvalidOperationException("Embedding dimensions do not match.");

        double dot = 0;
        double magA = 0;
        double magB = 0;

        for (var i = 0; i < a.Length; i++)
        {
            dot += a[i] * b[i];
            magA += a[i] * a[i];
            magB += b[i] * b[i];
        }

        if (magA == 0 || magB == 0)
            return 0;

        return dot / (Math.Sqrt(magA) * Math.Sqrt(magB));
    }
}
```

```csharp
public sealed class AdaptiveContextCompressor : IContextCompressor
{
    private readonly ILlmBackend _backend;
    private readonly IMemoryStore _memoryStore;

    public AdaptiveContextCompressor(
        ILlmBackend backend,
        IMemoryStore memoryStore)
    {
        _backend = backend;
        _memoryStore = memoryStore;
    }

    public async Task<CompressionResult> CompressAsync(
        ChatSession session,
        CompressionPolicy policy,
        CancellationToken ct = default)
    {
        var estimatedTokens = EstimateTokens(session);
        var usage = estimatedTokens / (double)policy.MaxContextTokens;

        if (usage < policy.LightThreshold)
        {
            return new CompressionResult(
                false,
                "",
                session.Messages,
                Array.Empty<string>());
        }

        var preserved = session.Messages
            .TakeLast(8)
            .ToList();

        var foldable = session.Messages
            .Take(Math.Max(0, session.Messages.Count - preserved.Count))
            .ToList();

        var residue = await BuildResidueAsync(foldable, ct);

        await _memoryStore.SaveAsync(
            $"residue:{session.SessionId}:{DateTime.UtcNow:yyyyMMddHHmmss}",
            residue,
            ct);

        return new CompressionResult(
            true,
            residue,
            preserved,
            Array.Empty<string>());
    }

    private static int EstimateTokens(ChatSession session)
    {
        var chars = session.Messages.Sum(x => x.Content.Length);
        return Math.Max(1, chars / 4);
    }

    private async Task<string> BuildResidueAsync(
        IReadOnlyList<ChatMessage> messages,
        CancellationToken ct)
    {
        var compressionSession = new ChatSession();

        compressionSession.AddSystem("""
        Compress the following conversation into operational residue.
        Preserve:
        - active goals
        - constraints
        - decisions
        - unresolved tasks
        - named systems
        Remove:
        - filler
        - duplicates
        - resolved branches
        Output compact YAML.
        """);

        compressionSession.AddUser(string.Join(
            "\n\n",
            messages.Select(m => $"{m.Role}: {m.Content}")));

        var output = "";

        await foreach (var token in _backend.GenerateAsync(
            compressionSession,
            new GenerationOptions(
                Temperature: 0.2,
                TopP: 0.8,
                MaxTokens: 700),
            ct))
        {
            output += token;
        }

        return output;
    }
}
```

```csharp
public sealed class ResourceGovernor : IResourceGovernor
{

    public ResourceGovernor(IGpuMonitor gpuMonitor)
    {
        _gpuMonitor = gpuMonitor;
    }

    public async Task<ResourceDecision> EvaluateAsync(
        ResourceRequest request,
        CancellationToken ct = default)
    {
        var gpu = await _gpuMonitor.GetStateAsync(ct);

        if (!gpu.Available)
        {
            return new ResourceDecision(
                Allowed: request.IsCritical,
                Mode: request.IsCritical ? "allow_cpu_fallback" : "queue",
                Reason: "GPU unavailable.",
                MaxConcurrentAgents: request.IsCritical ? 1 : 0,
                CompressionRequired: true
            );
        }

        var freeRatio = gpu.FreeVramMb / (double)Math.Max(1, gpu.TotalVramMb);

        if (freeRatio < 0.10)
        {
            return new ResourceDecision(
                Allowed: false,
                Mode: "halt",
                Reason: "VRAM critical.",
                MaxConcurrentAgents: 0,
                CompressionRequired: true
            );
        }

        if (freeRatio < 0.20)
        {
            return new ResourceDecision(
                Allowed: request.IsCritical,
                Mode: request.IsCritical ? "allow_critical" : "queue",
                Reason: "VRAM constrained.",
                MaxConcurrentAgents: 1,
                CompressionRequired: true
            );
        }

        if (freeRatio < 0.40)
        {
            return new ResourceDecision(
                Allowed: true,
                Mode: "throttle",
                Reason: "VRAM moderate.",
                MaxConcurrentAgents: 2,
                CompressionRequired: true
            );
        }

        return new ResourceDecision(
            Allowed: true,
            Mode: "allow",
            Reason: "Resources healthy.",
            MaxConcurrentAgents: 4,
            CompressionRequired: false
        );
```

```csharp
public sealed class DynamicTensorCompiler : IDynamicTensorCompiler
{
    public Task<CompiledTensor> CompileAsync(
        TensorCompileRequest request,
        CancellationToken ct = default)
    {
        var entropy = EstimateEntropy(request.RawInput, request.ActiveContext);
        var routeTarget = SelectRouteTarget(entropy);

        var preserved = request.ActiveContext
            .Where(IsCritical)
            .Take(12)
            .ToList();

        var removed = request.ActiveContext
            .Where(x => !IsCritical(x))
            .Take(50)
            .ToList();

        var compressedPrompt = BuildCompressedPrompt(
            request.RawInput,
            preserved);

        var tensor = new CompiledTensor(
            TensorId: $"tensor_{Guid.NewGuid():N}",
            Intent: ExtractIntent(request.RawInput),
            Domain: ExtractDomain(request.RawInput),
            Regime: entropy > 0.80 ? "R4" : entropy > 0.55 ? "R3" : "R2",
            Entropy: entropy,
            Grounding: EstimateGrounding(request.ActiveContext),
            Authority: EstimateAuthority(request.RuntimeState),
            RecursionDepth: GetRecursionDepth(request.RuntimeState),
            PreservedContext: preserved,
            RemovedContext: removed,
            RouteTarget: routeTarget,
            CompressedPrompt: compressedPrompt
        );

        return Task.FromResult(tensor);
    }

    private static double EstimateEntropy(
        string input,
        IReadOnlyList<string> context)
    {
        var branchWords = new[]
        {
            "maybe", "possibly", "unknown", "ambiguous",
            "multiple", "recursive", "speculative"
        };

        var score = branchWords.Count(w =>
            input.Contains(w, StringComparison.OrdinalIgnoreCase)) * 0.12;

        score += Math.Min(0.40, context.Count / 100.0);

        return Math.Clamp(score, 0.0, 1.0);
    }

    private static bool IsCritical(string value)
    {
        return value.Contains("id", StringComparison.OrdinalIgnoreCase)
            || value.Contains("constraint", StringComparison.OrdinalIgnoreCase)
            || value.Contains("deadline", StringComparison.OrdinalIgnoreCase)
            || value.Contains("hash", StringComparison.OrdinalIgnoreCase)
            || value.Contains("source", StringComparison.OrdinalIgnoreCase)
            || value.Contains("system", StringComparison.OrdinalIgnoreCase);
    }

    private static string SelectRouteTarget(double entropy)
    {
        if (entropy >= 0.80)
            return "qelf_tunnel";

        if (entropy >= 0.55)
            return "llm_backend_with_verification";

        return "llm_backend";
    }

    private static string BuildCompressedPrompt(
        string input,
        IReadOnlyList<string> preserved)
    {
        return $"""
        INTENT_INPUT:
        {input}

        PRESERVED_CONTEXT:
        {string.Join("\n", preserved)}

        EXECUTION_RULE:
        Use preserved context only.
        Do not restore removed context unless explicitly requested.
        """;
    }

    private static string ExtractIntent(string input)
    {
        return input.Length > 120 ? input[..120] : input;
    }

    private static string ExtractDomain(string input)
    {
        if (input.Contains("code", StringComparison.OrdinalIgnoreCase))
            return "coding";

        if (input.Contains("legal", StringComparison.OrdinalIgnoreCase))
            return "legal";

        if (input.Contains("drone", StringComparison.OrdinalIgnoreCase))
            return "drone";

        return "system";
    }

    private static double EstimateGrounding(IReadOnlyList<string> context)
    {
        return context.Any(x => x.Contains("source", StringComparison.OrdinalIgnoreCase))
            ? 0.80
            : 0.40;
    }

    private static double EstimateAuthority(IReadOnlyDictionary<string, object> state)
    {
        return state.ContainsKey("authority") ? 0.90 : 0.50;
    }

    private static int GetRecursionDepth(IReadOnlyDictionary<string, object> state)
    {
        return state.TryGetValue("recursion_depth", out var value) &&
```

```csharp
public sealed class GovernedExecutionRouter
{
    private readonly IDynamicTensorCompiler _compiler;
    private readonly IResourceGovernor _governor;

    public GovernedExecutionRouter(
        IDynamicTensorCompiler compiler,
        IResourceGovernor governor)
    {
        _compiler = compiler;
        _governor = governor;
    }

    public async Task<ExecutionRouteDecision> RouteAsync(
        TensorCompileRequest request,
        CancellationToken ct = default)
    {
        var tensor = await _compiler.CompileAsync(request, ct);

        var resourceDecision = await _governor.EvaluateAsync(
            new ResourceRequest(
                RequestId: request.RequestId,
                AgentType: tensor.RouteTarget,
                EstimatedContextTokens: tensor.CompressedPrompt.Length / 4,
                EstimatedVramMb: EstimateVram(tensor),
                CurrentRecursionDepth: tensor.RecursionDepth,
                IsCritical: tensor.Authority > 0.85),
            ct);

        return new ExecutionRouteDecision(
            tensor,
            resourceDecision,
            resourceDecision.Allowed
                ? tensor.RouteTarget
                : resourceDecision.Mode
        );
    }

    private static int EstimateVram(CompiledTensor tensor)
    {
        var baseCost = 512;
        var contextCost = tensor.CompressedPrompt.Length / 100;
        var recursionCost = tensor.RecursionDepth * 256;

        return baseCost + contextCost + recursionCost;
    }
```
